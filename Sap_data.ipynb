```python
import pandas as pd
import numpy as np

np.random.seed(42)
equipment_master = pd.read_csv("data/sap_equipment_master.csv")
machine_ids = equipment_master["Equipment_ID"].tolist()

# generate 180 days of daily production records per machine
date_range = pd.date_range(start="2025-01-01", periods=180, freq="D")

records = []
for machine in machine_ids:
    for date in date_range:
        planned_time = 480  
        ideal_cycle_time = np.random.uniform(0.8, 1.5)  # minutes per unit
        total_count = np.random.randint(250, 400)
        good_count = total_count - np.random.randint(0, 15)  # small scrap rate
        run_time = planned_time - np.random.randint(0, 60)  # some downtime eats into run time

        records.append({
            "Equipment_ID": machine,
            "Date": date,
            "Planned_Production_Time_min": planned_time,
            "Run_Time_min": run_time,
            "Ideal_Cycle_Time_min": round(ideal_cycle_time, 2),
            "Total_Count": total_count,
            "Good_Count": good_count
        })

sap_production = pd.DataFrame(records)
sap_production.to_csv("data/sap_production.csv", index=False)
print(sap_production.head())
print(sap_production.shape)
```

```python
np.random.seed(7)

downtime_reasons = ["Mechanical Failure", "Electrical Fault", "Tool Change",
                     "Material Shortage", "Operator Break", "Quality Check"]

downtime_records = []
for machine in machine_ids:
    # each machine has a random number of breakdown events over the 180 days
    n_events = np.random.randint(5, 20)
    event_dates = np.random.choice(date_range, size=n_events, replace=False)

    for date in event_dates:
        downtime_records.append({
            "Equipment_ID": machine,
            "Date": date,
            "Downtime_Reason": np.random.choice(downtime_reasons),
            "Downtime_Duration_min": np.random.randint(10, 120)
        })

sap_downtime = pd.DataFrame(downtime_records)
sap_downtime.to_csv("data/sap_downtime.csv", index=False)
print(sap_downtime.head())
print(sap_downtime.shape)
```
